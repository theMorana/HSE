For work with local models I chose Ollama

The task is to create multiple choice questions based on the received data (text fragments). The task was implemented as part of a summer internship.

Model - gemini 2:2b.

###Metrics results:

The accuracy of the phrasing (Precision = 0.1068) means that, on average, only 10.7% of the words from the generated question are found in the source text, that is, more than 89% of the vocabulary of the question is not related to the document.

The completeness of coverage (Recall = 0.0087) is even more critical – only 0.87% of the keywords in the source text are reflected in the question, which means that the content of the document is almost completely ignored.

Semantic proximity (SBERT = 0.2449) when normalizing cosine similarity (where 1 is identical meaning, 0 is orthogonality) shows a very low semantic overlap. Values above 0.7 are usually considered good, values above 0.8 are considered excellent, while a value of 0.24 indicates that the vector representations of the question and the source text are almost uncorrelated, meaning the questions are semantically far from the content.


###The selected models demonstrated poor quality of question generation for several reasons:

* complexity of technical texts
* the abundance of information in the data (formulas and tables in LaTex format created additional difficulties for the model)
* it is not possible to set the necessary parameters (for such a task, a lower temperature of the model is needed)
